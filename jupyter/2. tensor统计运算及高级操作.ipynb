## 一、统计分布

```python
from torch.distributions import Normal
```

```python
torch.manual_seed(1)
dist=Normal(loc=0,scale=1)  #正太分布，distributions中有很多种分布，不一一列举
sample=dist.sample((3,))  #随机采样。输入维度数量
log_prob=dist.log_prob(sample)  #计算对数概率
sample,log_prob
```

```python
torch.dist(sample,log_prob,p=2) #计算p范数。p=1是曼哈顿距离，2是欧式距离2范数
```

```python
torch.sum((sample-log_prob)**2).sqrt()
```

```python
torch.norm(sample) #直接计算某个tensor的2范数。对元数值的平方和进行开方
```

```python
torch.sum((sample)**2).sqrt()
```

## 三、tensor裁剪：

## 四、筛选填充操作

```python
nz=torch.nonzero(x)  #输出非0元素坐标
nz
```

```python
torch.take(x,torch.tensor([0,1,3,4,5,6,7]))  #将输入x视为1维数据（拉平），然后根据坐标获取相应元素
```

```python
one=torch.ones(x.shape) #torch.ones_like(x)
one
```

```python
cd=torch.where(x != 0,x,one)  #对x进行条件筛选，符合条件返回第一个x的值，不符合条件返回one的值
cd
```

```python
id=torch.index_select(x,dim=0,index=torch.tensor([2,0]))  #将第dim维的数据按照index顺序返回
id
```

```python
mask=x.gt(9)
mask
```

```python
torch.masked_select(x,mask)  #根据mask矩阵筛选True对应的元素并返回
```

## 五、拼接操作

```python
x=torch.randn(2,3)
x
```

```python
y=torch.randn(2,3)
y
```

### 5.1 拼接：维度不会变化

```python
torch.cat((x,y),dim=0)  #在0维上进行拼接
```

```python
torch.cat((x,y),dim=1)  #在0维上进行拼接
```

### 5.2 堆叠：会升维

```python
torch.stack((x,y),dim=0)  #在0维上进行拼接
```

```python
torch.stack((x,y),dim=1)  #在1维上进行拼接
```

```python
torch.stack((x,y),dim=2)  #在2维上进行拼接
```

## 六、切片（矩阵分块）

```python
a=torch.randn(9,4)
a
```

```python
torch.chunk(a,2,dim=0)  #在0维上切片，无法整除将导致最后一个tensor的数量小于均值（余数）
```

```python
torch.chunk(a,2,dim=1)
```

```python
torch.split(a,2,dim=0)   #split和chunk功能相似，但传入的第二个参数是维度size。这里值0维每2个元素切分一次
```

```python
torch.split(a,[3,6],dim=0)  #按照list中的size切分，list的和必须等于维度的size
```

## 七、变形reshape和转换

```python
a=torch.randn(3,4, 2)
a
```
