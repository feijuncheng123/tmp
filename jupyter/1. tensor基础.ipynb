## 基本运算

```python
x+y #或者torch.add(x,y)
x-y #或者torch.sub(x,y)
x * y #对应元素乘积。或者torch.mul(x,y) 
x / y #对应元素。或者torch.div(x,y) 

z=y.view(3,5)
z=z.type(torch.float64)
x @ z #矩阵点积。或者的torch.mm(x,z), x.mm(z)，x.matmul(z)等
x ** 2 # x.pow(2)
torch.exp(x) #自然常数e的x次方
x.sqrt() #开方。负值开方会转为nan
torch.log(x) #底数为e
torch.log2(x)
torch.log10(x)
```

</font>

```python
a1=torch.randn(3,2,2)
a1
```

```python
a2=torch.randn(2)
a2
```

# 反向传播

```python
x=torch.randn([3,4],requires_grad=True)
x
```

```python
y=torch.randn([3,4],requires_grad=True)
y
```

```python
z= x + y ** 2
z
```

```python
a=torch.randn([3,4],requires_grad=True)
a
```

```python
b = torch.sigmoid(z) + torch.sigmoid(a)
b
```

```python
result = b.sum()
result
```

<font color='green'>
<h2>反向传播</h2>
1. 调用backward方法，持续调用会梯度累计。<br />
2. 梯度传播公式f(g(x))=f'(g) * g'(x).
</font>

```python
result.backward()
```

<font color='green'>
3. result对x的梯度经过 result 对 z的梯度，又z对x的梯度传播。result 对 z的梯度 是sigmoid的梯度，z对x的梯度是 1 <br />
4. 因此x的梯度等于result关于z的梯度 乘与 z关于x的梯度1
</font>

```python
x.grad
```

<font color='green'>
5. sigmoid函数的导数,a是单层函数，没有经过其他层传播
</font>

```python
a.grad
```

<font color='green'>
6. y的梯度等于f'(z) * z'(y). 其中f'(z)是sigmoid函数对z值的导数，z'(y)是z= x + y ** 2中对的偏导数2y. 根据具体的z值求出f'(z)和2y求出z'(y)，相乘既是y的梯度
</font>

```python
y.grad
```
